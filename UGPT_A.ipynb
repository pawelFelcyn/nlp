````python
#@title dane

vocab = [
    "kot", "pies", "ptak", "ryba", "koń", "mysz",
    "jabłko", "banan", "chleb", "mleko", "ser", "mięso",
    "dom", "okno", "drzwi", "krzesło", "stół", "szafa",
    "idzie", "biegnie", "śpi", "je", "pije", "czyta",
    "zielony", "czerwony", "mały", "duży", "ładny", "brzydki"
]

words_for_answer = ["kot", "pies", "ptak", "ryba", "koń", "mysz"]
````

````python
#@title code

def word_to_vector(word, vocab):
    return [1 if w == word else 0 for w in vocab]

def vector_to_word(vector, vocab):
    for i, val in enumerate(vector):
        if val == 1:
            return vocab[i]
    return None
````

````python
# @title Answer
final_answers = []
for word in words_for_answer:
    final_answers.append(str(word_to_vector(word, vocab)))
    final_answers.append(vector_to_word(word_to_vector(word, vocab), vocab))
final_answer = ";".join(final_answers)
````

To metoda, która pozwala modelowi „nauczyć się” fragmentów słów (subwordów) poprzez łączenie najczęściej występujących par znaków.

---

## 🎯 Cel
Zrozumieć, jak działa **iteracyjne łączenie par symboli** i jak w praktyce powstaje słownik subwordowy.

---

## 📘 Kontekst
W tradycyjnych modelach każde słowo miało własny token — to prowadziło do problemu rzadkich słów i ogromnych słowników.  
BPE rozwiązuje to, **dzieląc słowa na znaki**, a następnie **łącząc najczęściej występujące pary** w nowe symbole.

---

## 🧱 Przykład działania (intuicyjnie)

```

```

2. Znajdź najczęstszą parę znaków — np. ('e','s')  
3. Połącz ją w jeden symbol: `es`
4. Zastąp wszystkie wystąpienia tej pary.

````python
#@title dane


words = [
    # rodzina "dom"
    "dom", "domek", "domy", "domowa", "domowy", "domkiem", "domkach", "domator",
    # rodzina "kot"
    "kot", "koty", "kotka", "kotki", "kotek", "kociak", "kocica", "koci",
    # rodzina "pies"
    "pies", "psa", "psy", "psiem", "psów", "psiaki", "psinka", "psinka",
    # rodzina "czytać"
    "czyta", "czytam", "czytasz", "czytał", "czytała", "czytali", "czytanie", "czytanka",
    # rodzina "pić / jeść"
    "je", "jem", "jesz", "jadł", "jadła", "jadłem", "pije", "pił"
]
````

````python
#@title code
def tokenize_word(vocab, word):
    tokens = []
    token = ''
    for letter in word:
        token += letter
        if not token in vocab:
            tokens.append(token[:-1])
            token = letter
    return tokens
        

def make_vocab(words):
    vocab = set([w for w in words])
    print(len(vocab))
    for _ in range(30):
        counts = {}
        for word in words:
            tokens = tokenize_word(vocab, word)
            for i in range(len(tokens)-1):
                pair = f'{tokens[i]}{tokens[i+1]}'
                if pair in counts:
                    counts[pair] += 1
                else:
                    counts[pair] = 1
        most_frequent = max(counts, key=counts.get)
        vocab.add(most_frequent)
    return vocab

vocab = make_vocab(words)
````

### 🧠 Pytania do refleksji

````python
# @title Answer
import sentencepiece as spm

spm.SentencePieceTrainer.Train(
    input='korpus.txt',
    model_prefix='demo_spm',
    vocab_size=100,
    character_coverage=1.0,
    model_type='unigram'
)
sp = spm.SentencePieceProcessor()
sp.Load('demo_spm.model')

sentences_from_corpus = ["słoń stoi w zoo", "kot śpi na kanapie", "pies biega po trawie"]
sentences_from_outside = ["tego zdania nie ma w korpusie", "to jest jakieś nowe zdanie"]

final_dictionary = []
for sentence in sentences_from_corpus + sentences_from_outside:
    tokens = sp.EncodeAsPieces(sentence)
    print(f'Sentence: {sentence}')
    print(f'Tokens: {tokens}')
    final_dictionary.append(f"{sentence}: {','.join(tokens)}")
final_answer = ";".join(final_dictionary)
print(final_answer)
````
